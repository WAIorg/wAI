```python
import open3d as o3d

mesh_path = "/Users/adeleyounis/Desktop/Capstone/wAI/volume-model/output/alexis_rgb/alexis_rgb.obj"
mesh = o3d.io.read_triangle_mesh(mesh_path)
mesh.compute_vertex_normals()
# o3d.visualization.draw_geometries([mesh])

bbox = mesh.get_axis_aligned_bounding_box()
ext = bbox.get_extent()
h_mesh = float(ext[2]) if ext[2] == max(ext) else float(max(ext))  # pick tallest axis
print("Mesh extent (m):", ext, " -> assumed height:", h_mesh, "m")

# Convert mesh to a voxel grid
voxel_size = 0.005  # adjust this for precision vs. speed
voxels = o3d.geometry.VoxelGrid.create_from_triangle_mesh(mesh, voxel_size=voxel_size)

# Compute volume estimate
voxel_volume = len(voxels.get_voxels()) * (voxel_size ** 3)

volume_litres = voxel_volume * 1000
print(f"Volume: {voxel_volume:.4f} m³ ({volume_litres:.2f} L)")
```
